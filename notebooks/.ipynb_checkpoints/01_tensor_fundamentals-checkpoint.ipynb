````python
import torch
import time

def benchmark_matmul(size):
    """Benchmark matrix multiplication at given size"""
    # Create tensors
    cpu_tensor = torch.randn(size, size)
    gpu_tensor = cpu_tensor.to("mps")
    
    # Warmup GPU
    _ = gpu_tensor @ gpu_tensor
    torch.mps.synchronize()
    
    # Time CPU
    start = time.time()
    _ = cpu_tensor @ cpu_tensor
    cpu_time = time.time() - start
    
    # Time GPU
    start = time.time()
    _ = gpu_tensor @ gpu_tensor
    torch.mps.synchronize()
    gpu_time = time.time() - start
    
    return cpu_time, gpu_time

# Test different sizes
sizes = [10000, 50000, 100000, 200000, 500000, 800000]

print("Size\t\tCPU Time\tGPU Time\tSpeedup")
print("-" * 60)

for size in sizes:
    cpu_t, gpu_t = benchmark_matmul(size)
    speedup = cpu_t / gpu_t
    print(f"{size}\t\t{cpu_t:.4f}s\t\t{gpu_t:.4f}s\t\t{speedup:.2f}x")
````

````python
### G P U Performance Hunt

# Find the crossover point where GPU becomes faster than CPU
# Try sizes: 100, 500, 1000, 2000, 5000, 8000
# Create a simple table showing CPU time vs GPU time
import torch
import time

def benchmark_matmul(size): #creating a function (reusable) to copy code n times when written once
    #CRETE TENSORS
    cpu_tensor = torch.randn(size,size) #create a tensor to generate random numbers ["random normal distribution"] per defined size
    gpu_tensor = cpu_tensor.to("mps") #create gpu tensor that copies cpu tensor

    #NOW WARMUP THE GPU
    _=gpu_tensor @ gpu_tensor # @-MATRIX MULTIPLICATION | _ -IDC BOUT THE RESULTS | 
                                #core operation in neural networks. When you train models, you're doing millions of these!

    
    torch.mps.synchronize() # SYNC()- waits for GPU to finish since GPU runs async

    #TIME CPU
    start = time.time() #Record start time
    _ = cpu_tensor @ cpu_tensor #Doing the computation here 
    cpu_time = time.time() #Calculate end-start , total duration ie

    #TIME GPU
    start = time.time() #Record start time
    _ = gpu_tensor @ gpu_tensor #computation
    torch.mps.synchronize() #sync()
    gpu_time = time.time() - start #################################

    return cpu_time, gpu_time

    #TEST DIFFERENT SIZES
    sizes = [100,200,1000,4000,5000,8000,10000]

    print("Size\t\t CPU time\t\t GPU Time \tSpeedup")
    print("-" * 60)

    for size in sizes:
    cpu_t, gpu_t = benchmark_matmul(size)
    speedup = cpu_t / gpu_t
    print(f"{size}\t\t{cpu_t:.4f}s\t\t{gpu_t:.4f}s\t\t{speedup:.2f}x")
````

````python
import torch
import time

def benchmark_matmul(size):
    """Benchmark matrix multiplication at given size"""
    # Create tensors
    cpu_tensor = torch.randn(size, size)
    gpu_tensor = cpu_tensor.to("mps")
    
    # Warmup GPU
    _ = gpu_tensor @ gpu_tensor
    torch.mps.synchronize()
    
    # Time CPU
    start = time.time()
    _ = cpu_tensor @ cpu_tensor
    cpu_time = time.time() - start
    
    # Time GPU
    start = time.time()
    _ = gpu_tensor @ gpu_tensor
    torch.mps.synchronize()
    gpu_time = time.time() - start
    
    return cpu_time, gpu_time

# Test different sizes
sizes = [100, 500, 1000, 2000, 5000, 8000]

print("Size\t\tCPU Time\tGPU Time\tSpeedup")
print("-" * 60)

for size in sizes:
    cpu_t, gpu_t = benchmark_matmul(size)
    speedup = cpu_t / gpu_t
    print(f"{size}\t\t{cpu_t:.4f}s\t\t{gpu_t:.4f}s\t\t{speedup:.2f}x")
````

````python
# What if we do MANY operations in sequence on GPU?
large_image_cpu = torch.randn(5000, 5000)
large_image_gpu = large_image_cpu.to("mps")

# CPU: normalize + square + sum
start = time.time()
result_cpu = ((large_image_cpu - large_image_cpu.mean()) / large_image_cpu.std()) ** 2
result_cpu = result_cpu.sum()
cpu_time = time.time() - start

# GPU: same operations
torch.mps.synchronize()
start = time.time()
result_gpu = ((large_image_gpu - large_image_gpu.mean()) / large_image_gpu.std()) ** 2
result_gpu = result_gpu.sum()
torch.mps.synchronize()
gpu_time = time.time() - start

print(f"CPU chain: {cpu_time:.6f}s")
print(f"GPU chain: {gpu_time:.6f}s")
print(f"Speedup: {cpu_time/gpu_time:.2f}x")
````

````python
import torch

#starting guess
guess = torch.tensor(3.0, requires_grad=True)
target= torch.tensor(5.0)


print(f"Starting guess: {guess.item()}")

learning_rate = 0.1 #how big of a step to take

#train for 10 steps
for step in range(10):
    loss = (guess- target) **2
    
    #cal gradient
    loss.backward()

    #Update guess (OPPOSITE direction of gradient!)
    with torch.no_grad(): #Dont track this update operation
        guess = guess - learning_rate * guess.grad
        guess.requires_grad = True #Re-enable gradient tracking

        #Clear gradients for next iteration
        guess.grad = None
 
        print(f"Step {step+1}: Guess = {guess.item():.4f},Loss = {loss.item():.4f}")

print(f"\nFinal guess: {guess.item():.4f}")
print(f"Target was: {target.item()}")

    
````

````python
import torch
import matplotlib.pyplot as plt

guess = torch.tensor(3.0, requires_grad=True)
target = torch.tensor(5.0)
learning_rate = 0.5

guesses = [guess.item()]
losses = []

for step in range(20):  # More steps to see the curve
    loss = (guess - target) ** 2
    losses.append(loss.item())
    
    loss.backward()
    
    with torch.no_grad():
        guess = guess - learning_rate * guess.grad
        guess.requires_grad = True
    
    guesses.append(guess.item())
    guess.grad = None

# Plot
plt.figure(figsize=(12, 4))

plt.subplot(1, 2, 1)
plt.plot(guesses, marker='o')
plt.axhline(y=5.0, color='r', linestyle='--', label='Target')
plt.xlabel('Step')
plt.ylabel('Guess Value')
plt.title('Learning Progress')
plt.legend()
plt.grid(True)

plt.subplot(1, 2, 2)
plt.plot(losses, marker='o', color='orange')
plt.xlabel('Step')
plt.ylabel('Loss')
plt.title('Loss Over Time')
plt.grid(True)

plt.tight_layout()
plt.show()

print(f"Started at: {guesses[0]:.2f}")
print(f"Ended at: {guesses[-1]:.4f}")
print(f"Target: {target.item()}")
````
